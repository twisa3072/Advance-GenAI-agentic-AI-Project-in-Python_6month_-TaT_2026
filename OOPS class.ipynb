```python
# Special Method: A special method is a method that is defined inside a class and has a special meaning in Python. It is used to perform operations on the data members of the class. A special method can be called using the object of the class.
```

```python
class computer():        #Didn't call the output   (User Defined Method)
    def config(self):      
        print('i5', '8gb')

com1 = computer()
```

```python
class computer():        
    def __init__(self):      
        print('i5', '8gb')

com1 = computer()
com2 = computer()
```

```python
class computer():        
    
    def __init__(self, cpu, ram, price): #__init__ means to initialize the variable
        print('in init')
    
    def config(self):     
        print('i5', '8gb',50000)

com1 = computer('i5', 8, 50000) # com1 has cpu i5 & ram is 8 
com2 = computer('i7', 16, 80000) # com2 has cpu intel 5 & ram is 8 as we have 2 computer we have different ram & cpt

com1.config() 
com2.config()
```

```python
class computer():        
     
    def __init__(self,cpu,ram): # cpu & ram as argument
        self.cpu = cpu #here self is the object & we need to refer to the object
        self.ram = ram 
 
    def config(self):     
        print('config is', self.cpu, self.ram) 

com1 = computer('i5', 8)
com2 = computer('i7', 16)


com1.config() 
com2.config()

#print(id(com1))
#print(id(com2))
```

```python
class computer():        
     
    def __init__(self,cpu,ram,price,company): # cpu & ram as argument
        self.cpu = cpu #here self is the object & we need to refer to the object
        self.ram = ram 
        self.price = price
        self.company = company

    def config(self):     
        print('config is', self.cpu, self.ram, self.price, self.company) 

com1 = computer('i5', 8, 50000, 'Lenevo')
com2 = computer('i7', 16, 80000, 'Hp')


com1.config() 
com2.config()

#print(id(com1))
#print(id(com2))
```
