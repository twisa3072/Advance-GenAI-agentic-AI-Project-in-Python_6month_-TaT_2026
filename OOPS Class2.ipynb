Instance Method --> The method which access all the instance variables.

```python
class student(): 
        
    school = 'nit'   
    
    def __init__(self, m1, m2, m3) : 
        self.m1 = m1
        self.m2 = m2
        self.m3 = m3

    def avg(self):   # ------> here this average is INSTANCE METHOD 
        return (self.m1+self.m2+self.m3)/3
    

s1 = student(10, 20, 30)
s2 = student(5, 15, 10)


print(s1.avg())
print(s2.avg())
```

Class Method

```python
class student(): 
         
    school = 'nit'
    
    def __init__(self, m1, m2, m3) :
        self.m1 = m1
        self.m2 = m2
        self.m3 = m3

    def avg(self):   # ------> hear this average is INSTANCE METHOD 
        return (self.m1+self.m2+self.m3)/3
    
    def info(cls): # user need to define cls - for class variable 
        return cls.school
s1 = student(10, 20, 30)
s2 = student(5, 15, 10)

print(s1.avg())
print(s2.avg())
print(s1.info())
```

Class Method using @classmethod decorator

```python
class student(): 
        
    school = 'nit'
    
    def __init__(self, m1, m2, m3) : # hear i pass 3 variable 
        self.m1 = m1
        self.m2 = m2
        self.m3 = m3

    def avg(self):   # ------> hear this average is INSTANCE METHOD 
        return (self.m1+self.m2+self.m3)/3
    
    @classmethod
    def info(cls): # user need to define cls - for class variable 
        return cls.school 
        

s1 = student(10, 20, 30)
s2 = student(5, 15, 10)

print(student.info()) 
print(s2.avg())
print(s1.avg())
```

Single Level Inheritence

```python
class A(): 
    def feature1(self):         
        print('feature1 is hi')
    def feature2(self):
        print('feature2 is hi')
        
class B(A):  #------>child class calling parent class...
    def feature3(self):
        print('feature3 is hello')
    def feature4(self):
        print('feature4 is hello')
a1 = A()  
b1 = B() 
a1.feature1()
a1.feature2()
b1.feature1()
b1.feature2()
b1.feature3()
b1.feature4()

```

Multi Level Inheritence , Whether to inherit the class or not its on our wish...like in this example I didn't inherit in class B.

```python
class A(): 
    def feature1(self):         
        print('feature1 is hi')
    def feature2(self):
        print('feature2 is hi')
        
class B():  
    def feature3(self):
        print('feature3 is hello')
    def feature4(self):
        print('feature4 is hello')
        
class C(A):
    def feature5(self):
        print('feature5 is welcome')

a1 = A()  
b1 = B()
c1 = C()

c1.feature1()
c1.feature2()
c1.feature5()
```

2nd Example of multilevel inheritence

```python
class A(): 
    def feature1(self):         
        print('feature1 is hi')
    def feature2(self):
        print('feature2 is hi')
        
class B(A):  
    def feature3(self):
        print('feature3 is hello')
    def feature4(self):
        print('feature4 is hello')
        
class C(B):
    def feature5(self):
        print('feature5 is welcome')

a1 = A()  
b1 = B()
c1 = C()

a1.feature1()
a1.feature2()
b1.feature1()
b1.feature2()
b1.feature3()
b1.feature4()
c1.feature1()
c1.feature2()
c1.feature3()
c1.feature4()
c1.feature5()
```

Multiple Inheritence

```python
class A(): 
    def feature1(self):         
        print('feature1 is hi')
    def feature2(self):
        print('feature2 is hi')
        
class B():
    def feature3(self):
        print('feature3 is hello')
    def feature4(self):
        print('feature4 is hello')

class C(A, B):
    def feature5(self):
        print('feature5 is welcome')

a1 = A()
b1 = B()
c1 = C()

c1.feature1()
c1.feature2()
c1.feature3()
c1.feature4()
c1.feature5()
```
